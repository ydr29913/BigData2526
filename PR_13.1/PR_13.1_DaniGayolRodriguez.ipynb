## Apartado A

1.  Realiza un programa en Python que a partir de los datos devueltos por la consulta nos muestre por pantalla los nombres de las categorías y el total de las mismas

```python
import requests

# Endpoint de las categorias
url = "https://www.thecocktaildb.com/api/json/v1/1/list.php?c=list"

response = requests.get(url)
data = response.json()

# Extraer las categorías
categorias = [item['strCategory'] for item in data['drinks']]

# Mostrar categorías
for categoria in categorias:
    print(f"- {categoria}")

# Total
print(f"\nTotal de las categorias: {len(categorias)}")
```

## Apartado B

1. Por cada una de las categorías genera un archivo JSON con el nombre de la misma que contenga todos sus cócteles

```python
import json

for categoria in categorias:
    # Sustituir los espacios en blanco por barras bajas
    categoria_limpia = categoria.replace(' ', '_').replace('/', '_')
    url = f"https://www.thecocktaildb.com/api/json/v1/1/filter.php?c={categoria.replace(' ', '_')}"
    
    response = requests.get(url)
    data = response.json()
    
    bebidas = data['drinks']
    
    # Guardar en un archivo JSON
    with open(f"{categoria_limpia}.json", "w", encoding="utf-8") as file:
        json.dump(bebidas, file, indent=4)
```

2. Según vas generando los archivos, cuenta los cócteles que contiene cada categoría mostrando al final del programa por pantalla el número total de cócteles por categorías

```python
import json

contar_categorias = {}

for categoria in categorias:
    # Sustituir los espacios en blanco por barras bajas
    url = f"https://www.thecocktaildb.com/api/json/v1/1/filter.php?c={categoria.replace(' ', '_')}"
    
    response = requests.get(url)
    data = response.json()
    
    bebidas = data['drinks']
    
    # Contar los cocteles
    contar_categorias[categoria] = len(bebidas)

# Mostrar los resultados
for categoria, total in contar_categorias.items():
    print(f"- {categoria}: {total}")
```

3. Almacena todos los datos de los cócteles en un único archivo con formato csv que contendrá para cada cóctel cuatro campos: su Id, nombre, imagen y categoría. La primera fila del archivo ha de incorporar el nombre de los campos

```python
import csv

with open("bebidas.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    
    # Nombre de los campos
    writer.writerow(["id", "nombre", "imagen", "categoria"])
    
    for categoria in categorias:
        # Sustituir los espacios en blanco por barras bajas
        url = f"https://www.thecocktaildb.com/api/json/v1/1/filter.php?c={categoria.replace(' ', '_')}"
        
        response = requests.get(url)
        data = response.json()
        
        bebidas = data['drinks']
        
        for bebida in bebidas:
            writer.writerow([
                bebida['idDrink'],
                bebida['strDrink'],
                bebida['strDrinkThumb'],
                categoria
            ])
```
